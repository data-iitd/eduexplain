```python
from transformers import Qwen2_5_VLForConditionalGeneration, AutoProcessor
import torch

model_name = "Qwen/Qwen2.5-VL-3B-Instruct"

model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

processor = AutoProcessor.from_pretrained(model_name)

print("Model loaded")
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
from PIL import Image
import torch

def generate_description(image_path):
    image = Image.open(image_path).convert("RGB")

    prompt = """
You are an expert science teacher.

Describe the given textbook figure in detail in 4-5 sentences.

ONLY output the description text.
Do NOT include labels, headings, or anything else.
"""

    messages = [
        {
            "role": "user",
            "content": [
                {"type": "image", "image": image},
                {"type": "text", "text": prompt}
            ]
        }
    ]

    text = processor.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = processor(
        text=[text],
        images=[image],
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=200
        )

    # ✅ Decode output
    result = processor.batch_decode(output, skip_special_tokens=True)[0]

    # 🔥 Extract ONLY assistant response
    if "assistant" in result:
        result = result.split("assistant")[-1]

    # 🔥 Remove unwanted labels
    result = result.replace("Description:", "")

    # 🔥 Clean formatting
    result = result.replace("\n", " ").strip()

    return result
```

```python
# Run description on all uploaded images

for image_name in uploaded.keys():
    print(f"\nProcessing: {image_name}")

    description = generate_description(image_name)

    print("\n📝 Description:")
    print(description)
```
